## Variablar og 'er lik'-teikn i Python

I Python kan vi opprette noko vi kallar variablar. 

Vi kan tenkje på ein variabel som ein boks. Boksen har eit namn, og oppi boksen kan vi legge 'ting'. Slike 'ting' kallar vi ofte verdiar.

**Døme:** I cella (ruta) under skjer det to ting. 

For det fyrste opprettar vi ein variabel som har namnet `sidelengd`. 

For det andre puttar vi verdien `3` oppi denne variabelen. 

Vi seier at vi lagrar verdien `3` i variabelen `sidelengd`.

```python
sidelengd = 3
```

No har vi sett korleis vi kan opprette variablar i Python. Men kva skal vi med dette? 

Ein viktig fordel med variablar er det fylgjande: Når vi fyrst har oppretta ein variabel og lagra ein verdi i denne variabelen, så veit datamaskina at neste gong vi skriv variabelnamnet, så vil vi eigentleg at datamaskina skal tenkje på verdien til variabelen. 

**Døme:** 

(Køyr cella under ved å setje markøren inni og klikke på 'play'-knappen.)

```python
sidelengd = 3

areal = sidelengd**2

print("Arealet av eit kvadrat der sidene har lengd")
print(sidelengd)
print("er")
print(areal)
```

Når vi køyrer programmet ovanfor, så tenkjer datamaskina slik: 
1) Dei vil opprette ein variabel som heiter `sidelengd`
2) I variabelen `sidelengd` vil dei lagre verdien `3`
3) Dei vil opprette ein variabel som heiter `areal`.
4) I variabelen `areal` vil dei lagre det talet vi får viss vi tek verdien av variabelen `sidelengd` og multipliserer denne med seg sjølv. Sidan `sidelengd` har verdien `3` så blir dette talet `9`. 
5) Dei vil at vi skal skrive ut tekststrengen `Arealet av eit kvadrat der sidene har lengd`
6) Dei vil at vi skal skrive ut verdien av variabelen `sidelengd`
7) Dei vil at vi skal skrive ut tekststrengen `"er"`
8) Dei vil at vi skal skrive ut verdien av variabelen `areal`

**Oppgåve:** 
Tenk på eit positivt tal. Prøv å endre programmet over slik at det reknar ut arealet av eit kvadrat som har sidelengder lik det positive talet du tenkte på.


### Variabelverdiar kan oppdaterast

Når vi har oppretta og gjeve verdi til ein variabel i Python, er det ikkje slik at variabelen må ha den same verdien for alltid. 

Ein variabelverdi kan oppdaterast. Vi skal no sjå på tre måtar dette kan gjerast.

**Døme:**

I cella under opprettar vi fyrst ein variabel som vi kallar `x`, og vi gjev variabelen verdien `4`. Dette skjer på linje 1. På linje 2 skriv vi ut verdien av variabelen `x`. Altså skriv vi ut `4`. 

Kva skjer eigentleg på linje 3 og 4? 

Tenk over kva du trur kjem til å bli printa ut, og køyr deretter cella for å sjekke kva som faktisk blir printa ut.

```python
x = 4
print(x)
x = 7 
print(x)
```

På linje 3 i cella over seier vi til datamaskina:

<div style="border: 2px solid #333; border-radius: 15px; padding: 12px; width: fit-content; background-color: #ffd966; color: #000000; box-shadow: 2px 2px 5px rgba(0,0,0,0.2);">
  Hallo datamaskin! Du, vi vil gjerne bytte ut innhaldet i boksen (variabelen) <code>x</code> frå den gamle verdien (altså <code>4</code>) til verdien <code>7</code>. Frå og med no vil vi at du skal tenkje på verdien <code>7</code> når vi skriv variabelnamnet <code>x</code>.
</div>

På linje 4 ber vi datamaskina om å printe den oppdaterte verdien av variabelen `x`.






**Døme:**

Ein annan måte å oppdatere verdien av ein variabel på er å fortelje datamaskina at vi vil at den nye variabelen skal vere svaret på eit reknestykke der den gamle variabelverdien inngår. 

Sjå på koden i cella under, tenk over kva du trur vil bli printa og sjekk forslaget ditt til svar ved å køyre koden i cella.

```python
y = 4
y = y + 1
print(y)
```

På linje 1 i cella over opprettar vi variabelen `y` og gjev han verdien `4`. 
På linje 2 seier vi til datamaskina: 

<div style="border: 2px solid #333; border-radius: 15px; padding: 12px; width: fit-content; background-color: #ffd966; color: #000000; box-shadow: 2px 2px 5px rgba(0,0,0,0.2);">
  Hallo datamaskin! Du, vi vil gjerne oppdatere verdien av variabelen <code>y</code>. Vi vil at 
    <span style="color:darkred">$\text{den nye variabelverdien}$</span> skal vere lik svaret på reknestykket 
    <div style="margin-left: 20px;">
    <span style="color:darkred">$\text{den nye variabelverdien}$</span>$ + 1$
  </div><br>
    Altså vil vi at <span style="color:darkred">$\text{den nye variabelverdien}$</span> skal vere éin meir enn <span style="color:darkred">$\text{den gamle variabelverdien}$</span>
</div>

Ein alternativ måte å skrive dette på er vist i cella under. Køyr koden i cella under og sjekk at utskrifta blir den same som frå førre kodecelle.

```python
y = 4
y += 1
print(y)
```

På linje 2 i cella over seier vi til datamaskina at vi vil at variabelverdien til variabelen `y` skal auke med $1$.
