## Datatypar

I Python finst det ulike datatypar som blir lagra og handsama på ulike måtar når datamaskina tolkar instruksjonane i programmet ditt. Her er nokre viktige datatypar: 

| Norsk namn           | Engelsk namn                  | Forklaring | Python-symbol | Døme          | Døme |
|----------------------|-------------------------------|------------|---------------|---------------|------|
| heilt tal            | integer                       |            | int           | 38            | -5   |
| flyttal              | floating point number (float) | desimaltal | float         | 4.673         | -5.0 |
| tekststreng (streng) | text string (string)          | tekst      | str           | 'Hello world' | "38" |

Ved hjelp av funksjonen type() kan vi sjekke datatypen til eit objekt, slik dømet under viser.
