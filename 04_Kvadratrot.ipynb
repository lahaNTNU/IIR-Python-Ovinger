## Kvadratrot

På engelsk heiter kvadratrot "square root". Når vi skal rekne ut kvadratrøter i Python kan vi bruke forkortinga `sqrt`. For å kunne gjere dette må vi hente ut funksjonen `sqrt()` frå modulen `math`. Vi kan rekne ut $\sqrt{9}$ og $\sqrt{8}$ slik:

```python
from math import sqrt

sqrt(9)
```

```python
from math import sqrt

sqrt(8)
```

Legg merke til at talet over er ein avrundingsverdi. Viss du skriv $\sqrt{8}$ på kalkulatoren, så kan du få den eksakte verdien $2\sqrt{2}$ som svar. Alternativt, på kalkulatoren, så kan du endre format og få ein avrundingsverdi som svar der også.
