## Talrekning

I Python kan vi gjere mange rekneoperasjonar. I døma under er det vist korleis du kan 
- addere (pluss),
- subtrahere (minus),
- multiplisere (gonge),
- bruke potensar,
- dividere (dele),
- finne resten ved divisjon og
- bruke parentesar og fleire rekneoperasjonar saman.

### Døme

(I alle kode-cellene under kan du køyre koden ved å setje markøren inne i cella og klikke på 'play'-knappen.)

Her adderer vi to tal: 

```python
5+4
```

Her subtraherer vi to tal:

```python
8-8
```

Her multipliserer vi to tal:

```python
2*3
```

Her finn vi potensen $4^3$, altså $4\cdot 4\cdot 4$

```python
# Vi reknar ut ein potens:
4**3
```

Her dividerer vi eit tal med eit anna: 

```python
15/3
```

Her finn vi resten ved divisjon av eit heilt tal med eit anna heilt tal:

```python
17%3
```

Viss du køyrer cella ovanfor ser du at resultatet av operasjonen 17%3 er 2. Grunnen til det er at $3\cdot 5+2=17$, så når du dividerer $17$ med $3$ blir resten $2$.

Vi kan også skrive uttrykk med fleire rekneoperasjonar på same linje. Då tolkar datamaskina det vi har skrive i tråd med dei same reglane for reknerekkjefylgje som vi bruker når vi reknar for hand.

```python
2*3+4
```

```python
(2*3)+4
```

```python
2*(3+4)
```
