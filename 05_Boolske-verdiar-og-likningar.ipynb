## Å setje prøve på svaret og boolske verdiar

I Python finst det ein enkel måte å sjekke om to tal er like. Dette kan vi bruke når vi har eit forslag til løysing av ei likning og vil setje prøve på svaret. 

Når vi skal sjekke om to tal er like, kan vi skrive `==` mellom dei to tala. 

Viss vi i tillegg bruker funksjonen `print()`, som i cellene under, så fortel datamaskina oss om det er sant (`True`) eller usant (`False`) at dei to tala er like.


```python
print( 5 == 5 )
```

```python
print( 5 == 6 )
```

```python
print( 4.6 == 4.8 )
```

```python
print( 4+2 == 6 )
```

```python
print( 4+2 == 4+3 )
```

Variablar som har éin av dei to mogelege verdiane `True` og `False` kallar vi boolske variablar. Desse to verdiane kallar vi boolske verdiar. 

Éin av dei tinga vi kan bruke boolske verdiar til er å setje prøve på svaret i likningar.

**Døme:**

I cella under sjekkar vi om $x=3$ er ei løysing av likninga 
&nbsp;&nbsp;&nbsp;&nbsp;$$2x+1=9-2.$$

```python
x = 3   # Forslag til løysing

venstre_side = 2*x+1         # Vi kan bruke _ som del av variabelnamn
hoegre_side = 9-2            # Vi unngår æ, ø og å i variabelnamn

# Vi sjekkar om det er sant eller usant at venstresida er lik høgresida:
print( venstre_side == hoegre_side ) 
```

Sidan `True`blir printa, og ikkje `False`, så veit vi at $x=3$ faktisk <u>er</u>
ei løysing av likninga. Det er sant at venstresida er lik høgresida når $x$ har verdien $3$. 

**Døme:**

Vi ser på éi likning til. I cella under sjekkar vi om $x=-6$ er ei løysing av likninga 
&nbsp;&nbsp;&nbsp;&nbsp;$$ \frac{x}{3}-4 = 2\left( 1+\frac{2x}{3} \right) $$

```python
x = -6 

venstre_side = x/3-4 
hoegre_side = 2*( 1 + (2*x)/3 )


# Vi sjekkar om det er sant eller usant at venstresida er lik høgresida:
print( venstre_side == hoegre_side ) 
```

Vi kan konkludere slik: Ja, det er sant at $x=-6$ er ei løysing av likninga.

Til slutt testar vi éin annan $x$-verdi for den same likninga. I cella under sjekkar vi om $x=5$ er ei løysing av likninga
&nbsp;&nbsp;&nbsp;&nbsp;$$ \frac{x}{3}+4 = 2\left( 1+\frac{2x}{3} \right) $$

```python
x = 9

venstre_side = x/3-4 
hoegre_side = 2*( 1 + (2*x)/3 )

# Vi sjekkar om det er sant eller usant at venstresida er lik høgresida:
print( venstre_side == hoegre_side ) 
```

Vi kan konkludere med at $x=5$ ikkje er ei løysing av likninga.
